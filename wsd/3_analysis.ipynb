```python
# en -> zh
def print_translations():
    with open("./en-zh.en-filtered-wsd-processed.en.subword.test.desubword", 'r', encoding='utf-8') as en_file, \
         open("./zh.translated.desubword", 'r', encoding='utf-8') as zh_bert_file, \
         open("./en-zh.zh-filtered-wsd.zh.subword.test.desubword", encoding="utf-8") as zh_correct_file:
        
        for en_line, zh_bert_line, zh_correct_line in zip(en_file, zh_bert_file, zh_correct_file):
            en_line = en_line.strip()
            zh_bert_line = zh_bert_line.strip()
            # zh_base_line = zh_base_line.strip()
            zh_correct_line = zh_correct_line.strip()
            
            print(f"English:            {en_line}")
            print(f"Chinese (bert):     {zh_bert_line}")
            # print(f"Chinese (base):     {zh_base_line}")
            print(f"Chinese (correct):  {zh_correct_line}")
            print("-" * 50)

print_translations()
```
